```python
s_day = int(input("Enter youre day in dd formate :"))
s_month = int(input('Enter youre starting month :'))
s_year = int(input("Enter youre sarting years :"))

e_day = int(input("Enter youre day in dd formate :"))
e_month = int(input('Enter youre Ending month :'))
e_year = int(input("Enter youre Ending years :"))

T_years = 0
month_count = 0
day_count = 0

# ૧. e_year + 1 કર્યું જેથી છેલ્લું વર્ષ પણ લૂપમાં ગણાય
for i in range(s_year, e_year + 1):
    if i == s_year:
        # ૨. ૧૩ કર્યું જેથી ૧૨મો મહિનો ગણાય
        for j in range(s_month, 13):
            # ૩. ૩૧ કર્યું જેથી ૩૦મો દિવસ ગણાય 
            for k in range(s_day, 31):
                day_count += 1
            # મહિનો પૂરો થાય એટલે કાઉન્ટ અહીં વધે, દિવસની અંદર નહીં
            month_count += 1
    
    # જો છેલ્લું વર્ષ હોય
    elif i == e_year:
        for j in range(1, e_month + 1):
            # છેલ્લા મહિનામાં માત્ર એન્ડિંગ દિવસ સુધી જ લૂપ ચલાવવી
            if j == e_month:
                for k in range(1, e_day + 1):
                    day_count += 1
            else:
                for k in range(1, 31):
                    day_count += 1
                month_count += 1
                
    else:
        for j in range(1, 13):
            for k in range(1, 31):
                day_count += 1
            month_count += 1
        T_years += 1

# વર્ષનો કાઉન્ટ સેટ કરવા માટે
T_years = month_count // 12

print("Years:", T_years)
print("Months:", month_count)
print("Days:", day_count)
```

```python
# યુઝર પાસેથી ઇનપુટ લેવું
s_day = int(input("Enter starting day (dd): "))
s_month = int(input("Enter starting month (mm): "))
s_year = int(input("Enter starting year (yyyy): "))

e_day = int(input("Enter ending day (dd): "))
e_month = int(input("Enter ending month (mm): "))
e_year = int(input("Enter ending year (yyyy): "))

# દરેક મહિનાના દિવસો (સામાન્ય વર્ષ માટે, ઇન્ડેક્સ 1 થી 12 માટે પહેલો 0 રાખ્યો છે)
days_in_months = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

total_days = 0
total_months = 0
total_years = 0

# વર્ષની ગણતરી માટે ટેમ્પરરી વેરિયેબલ
current_year = s_year
current_month = s_month
current_day = s_day

# જ્યાં સુધી આપણે એન્ડિંગ ડેટ પર ન પહોંચીએ ત્યાં સુધી એક-એક દિવસ આગળ વધો
while not (current_year == e_year and current_month == e_month and current_day == e_day):
    
    # લિપ વર્ષ (Leap Year) ચેક કરવા માટે (ફેબ્રુઆરીમાં 29 દિવસ કરવા)
    if (current_year % 4 == 0 and current_year % 100 != 0) or (current_year % 400 == 0):
        days_in_months[2] = 29
    else:
        days_in_months[2] = 28

    current_day += 1
    total_days += 1

    # જો દિવસ તે મહિનાના મહત્તમ દિવસો કરતાં વધી જાય, તો મહિનો બદલો
    if current_day > days_in_months[current_month]:
        current_day = 1
        current_month += 1
        total_months += 1

    # જો મહિનો 12 કરતાં વધી જાય, તો વર્ષ બદલો
    if current_month > 12:
        current_month = 1
        current_year += 1
        total_years += 1

# વર્ષ અને મહિનાની ચોક્કસ ગણતરી (વધારાના દિવસો બાદ કરીને)
# final_months = (e_year - s_year) * 12 + (e_month - s_month)
if e_day < s_day:
    # final_months -= 1
    total_months-=1
# final_years = final_months // 12

print("\n---  (Results) ---")
print(f" (Total Days): {total_days}")
print(f" (Total Months): {total_months}")
print(f" (Total Years): {total_years}")
```

```python
s_day = int(input("Enter starting day (dd): "))
s_month = int(input("Enter starting month (mm): "))
s_year = int(input("Enter starting year (yyyy): "))

e_day = int(input("Enter ending day (dd): "))
e_month = int(input("Enter ending month (mm): "))
e_year = int(input("Enter ending year (yyyy): "))

total_years = 0
total_months = 0
total_days = 0

# ગણતરી કરવા માટે તમારા ચલ (Variables) સેટ કર્યા
current_year = s_year
current_month = s_month
current_day = s_day

# જ્યાં સુધી આપણે એન્ડિંગ ડેટ (તારીખ, મહિનો, વર્ષ) સુધી ન પહોંચીએ ત્યાં સુધી લૂપ ચાલશે
while not (current_year == e_year and current_month == e_month and current_day == e_day):
    
    current_day += 1
    total_days += 1
    
    # ધારી લઈએ કે દરેક મહિનામાં ૩૦ દિવસ છે (તમારા જૂના લોજીક પ્રમાણે ફિક્સ ૩૦ દિવસ)
    if current_day > 30:
        current_day = 1
        current_month += 1
        total_months += 1
        
    # જો ૧૨ મહિના પૂરા થાય, તો વર્ષ બદલાશે
    if current_month > 12:
        current_month = 1
        current_year += 1

# કુલ મહિના અને વર્ષની ગણતરી લૂપ પૂરી થયા પછી
total_years = total_months // 12

print("\n--- પરિણામ (Results) ---")
print("Total Years :", total_years)
print("Total Months:", total_months)
print("Total Days  :", total_days)
```
